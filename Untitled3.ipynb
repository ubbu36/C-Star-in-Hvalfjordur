```python
import xarray as xr
import numpy as np
from roms_tools import Grid, ROMSOutput
# -----------------------------------
# Paths and grid configuration
# -----------------------------------
model_grid_path = "/home/x-uheede/S/Iceland2_MARBL_2024_60m/P_INPUT/Iceland2_grid.nc"

model_bgc_dia_path1 = (
    "/anvil/scratch/x-uheede/Iceland2_MARBL_2024_diags/"
    "Iceland2_MARBL_2024_bgc_dia_avg.2024??????????.nc"
)

target_depth_levels = [1,2,3,4,5,10,15,20,30,40]

# Grid parameters
vert_levels = 60
theta_s_model = 5
theta_b_model = 2
hc_model = 300

# -----------------------------------
# Create ROMS grid
# -----------------------------------
grid = Grid.from_file(
    model_grid_path
)

# -----------------------------------
# Create ROMSOutput object
# -----------------------------------
roms_bgc_dia1 = ROMSOutput(
    grid=grid,
    path=[model_bgc_dia_path1],
    use_dask=True,
)

# -----------------------------------
# Regrid ALL variables
# -----------------------------------
ds_bgc_dia1 = roms_bgc_dia1.regrid(
    depth_levels=target_depth_levels
)

# -----------------------------------
# Result
# -----------------------------------
ds_bgc_dia1
```

```python
CaCO3_PROD=ds_bgc_dia1['CaCO3_PROD'].sel(lon=slice(-21.6+360,-21.37+360)).sel(lat=slice(64.34,64.42)).mean('lon').mean('lat')

import matplotlib.pyplot as plt

# ensure proper ordering
CaCO3_PROD = CaCO3_PROD.transpose("time", "depth")

fig, ax = plt.subplots(figsize=(10, 5))

pcm = ax.pcolormesh(
    CaCO3_PROD.time,
    CaCO3_PROD.depth,
    CaCO3_PROD.T,
    shading="auto"
)

ax.invert_yaxis()
ax.set_ylabel("Depth (m)")
ax.set_xlabel("Time")
ax.set_title("CaCO3 Production Hovmöller")

cbar = plt.colorbar(pcm, ax=ax)
cbar.set_label("CaCO3_PROD")

plt.tight_layout()
plt.show()
```

```python
# Is this significant?

area = 35 * 4 * 10**6  # m²
time = 3600 * 24 * 30   # one month in seconds
average = 1 * 10**-5    # average flux rate

# Total sedimentation in month of May
sed_sum = area * time * average
print(f"sed sum = {sed_sum:.2e} mmol")

# Alkalinity loss
depth = 40          # m
alk_loss = 30       # mmol/m³
alk_loss_sum = depth * area * alk_loss
print(f"alk loss sum = {alk_loss_sum:.2e} mmol")

# Difference
difference = alk_loss_sum - sed_sum
print(f"difference = {difference:.2e}")
```
