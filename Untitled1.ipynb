```python
import matplotlib.pyplot as plt
import numpy as np
from roms_tools import Grid
from roms_tools.setup.mask import _close_narrow_channels, add_velocity_masks
```

```python
# Create a small grid with close_narrow_channels=False to avoid closing during init
grid = Grid(
    nx=15,
    ny=15,
    size_x=100,
    size_y=100,
    center_lon=-20,
    center_lat=64,
    rot=0,
    N=3,
    close_narrow_channels=False,
)

mask_shape = grid.ds["mask_rho"].shape
mask = np.zeros(mask_shape, dtype=np.int32)

# Create test mask: vertical ocean line, lower half ocean, small lake with narrow channel
line_xi = 8
mask[5:11, line_xi] = 1  # Vertical ocean line
mask[mask_shape[0] // 2 :, :] = 1  # Lower half ocean
mask[3:6, 4:7] = 1  # Small lake
mask[4, 6:8] = 1  # Narrow channel connecting lake to ocean

grid.ds["mask_rho"].values[:] = mask
mask_before = grid.ds.mask_rho.values.copy()

print(f"Mask shape: {mask_shape}")
print(f"Ocean cells before: {mask_before.sum()}")
print(f"Land cells before: {(mask_before == 0).sum()}")
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
im = ax.imshow(mask_before, cmap="Blues", vmin=0, vmax=1, origin="upper")
ax.set_title("Mask Before Closing Narrow Channels", fontsize=14, fontweight="bold")
ax.set_xlabel("xi_rho")
ax.set_ylabel("eta_rho")

# Highlight specific areas mentioned in the test
# Narrow channel at [6, 7]
ax.plot(7, 6, "ro", markersize=10, label="Channel [6,7]")
# Narrow channel at [6, 9]
ax.plot(9, 6, "ro", markersize=10, label="Channel [6,9]")
# Lake channel at [4, 7]
ax.plot(7, 4, "go", markersize=10, label="Lake channel [4,7]")
# Small lake region
from matplotlib.patches import Rectangle
rect_lake = Rectangle(
    (4, 3), 3, 3, linewidth=2, edgecolor="green", facecolor="none", linestyle="--"
)
ax.add_patch(rect_lake)
# Vertical ocean line
ax.axvline(x=line_xi, color="blue", linestyle=":", alpha=0.5, linewidth=2)

ax.legend(loc="upper right")
ax.grid(True, alpha=0.3)
plt.colorbar(im, ax=ax, label="Mask (0=Land, 1=Ocean)")
plt.tight_layout()
plt.show()
```

```python
# Close narrow channels
grid.ds = _close_narrow_channels(
    grid.ds,
    mask_var="mask_rho",
    max_iterations=10,
    inplace=True,
)
grid.ds = add_velocity_masks(grid.ds)

mask_after = grid.ds.mask_rho.values.copy()

print(f"Ocean cells after: {mask_after.sum()}")
print(f"Land cells after: {(mask_after == 0).sum()}")
print(f"\nCells filled (1→0): {((mask_after - mask_before) == -1).sum()}")
print(f"Cells opened (0→1): {((mask_after - mask_before) == 1).sum()}")
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
im = ax.imshow(mask_after, cmap="Blues", vmin=0, vmax=1, origin="upper")
ax.set_title("Mask After Closing Narrow Channels", fontsize=14, fontweight="bold")
ax.set_xlabel("xi_rho")
ax.set_ylabel("eta_rho")

# Highlight the same areas to show they've been closed
ax.plot(7, 6, "rx", markersize=12, markeredgewidth=2, label="Closed [6,7]")
ax.plot(9, 6, "rx", markersize=12, markeredgewidth=2, label="Closed [6,9]")
ax.plot(7, 4, "gx", markersize=12, markeredgewidth=2, label="Closed [4,7]")
# Show if lake was filled
if mask_after[3:6, 4:7].sum() == 0:
    rect_lake_filled = Rectangle(
        (4, 3),
        3,
        3,
        linewidth=2,
        edgecolor="red",
        facecolor="none",
        linestyle="--",
        label="Lake filled",
    )
    ax.add_patch(rect_lake_filled)

ax.legend(loc="upper right")
ax.grid(True, alpha=0.3)
plt.colorbar(im, ax=ax, label="Mask (0=Land, 1=Ocean)")
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Plot before
ax1 = axes[0]
im1 = ax1.imshow(mask_before, cmap="Blues", vmin=0, vmax=1, origin="upper")
ax1.set_title("Before", fontsize=14, fontweight="bold")
ax1.set_xlabel("xi_rho")
ax1.set_ylabel("eta_rho")
ax1.grid(True, alpha=0.3)
plt.colorbar(im1, ax=ax1, label="Mask (0=Land, 1=Ocean)")

# Plot after
ax2 = axes[1]
im2 = ax2.imshow(mask_after, cmap="Blues", vmin=0, vmax=1, origin="upper")
ax2.set_title("After", fontsize=14, fontweight="bold")
ax2.set_xlabel("xi_rho")
ax2.set_ylabel("eta_rho")
ax2.grid(True, alpha=0.3)
plt.colorbar(im2, ax=ax2, label="Mask (0=Land, 1=Ocean)")

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 8))
mask_diff = mask_after - mask_before
im = ax.imshow(
    mask_diff, cmap="RdBu_r", vmin=-1, vmax=1, origin="upper", interpolation="nearest"
)
ax.set_title("Difference (After - Before)", fontsize=14, fontweight="bold")
ax.set_xlabel("xi_rho")
ax.set_ylabel("eta_rho")
ax.text(
    0.5,
    0.02,
    "Red = Filled (1→0), Blue = Opened (0→1)",
    transform=ax.transAxes,
    ha="center",
    fontsize=10,
    bbox=dict(boxstyle="round", facecolor="wheat", alpha=0.5),
)
plt.colorbar(im, ax=ax, label="Change in Mask")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
