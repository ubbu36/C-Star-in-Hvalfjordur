```python
import xarray as xr
emod=xr.open_dataset('/anvil/projects/x-ees250129/Datasets/EMODnet_C2.nc')
```

```python
import matplotlib.pyplot as plt
plt.contourf(emod.lon, emod.lat, emod['elevation'])
```

```python
emod_coarse = emod.coarsen(
    lat=100,
    lon=100,
    boundary="trim"   # drops leftover cells at edges
).mean()
```

```python
emod_coarse.to_netcdf(
    "/anvil/projects/x-ees250129/x-uheede/EMODnet_C2_coarse100.nc"
)
```
